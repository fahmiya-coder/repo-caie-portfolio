```python
def is_even(n):
    """True if n is an even number, otherwise False."""
    return n % 2 == 0

def sum_to(n):
    """Add up 1 + 2 + 3 + ... + n."""
    total = 0
    for i in range(1, n + 1):
        total += i
    return total

def times_table(n, upto=10):
    """Return the n times table as a list of strings."""
    return [f"{n} x {i} = {n * i}" for i in range(1, upto + 1)]

def fizzbuzz(n):
    """Fizz for 3s, Buzz for 5s, FizzBuzz for both."""
    out = []
    for i in range(1, n + 1):
        if i % 15 == 0:
            out.append("FizzBuzz")
        elif i % 3 == 0:
            out.append("Fizz")
        elif i % 5 == 0:
            out.append("Buzz")
        else:
            out.append(str(i))
    return out

# test it
print(is_even(4), is_even(7))
print(sum_to(5))
print(times_table(3, 3))
print(fizzbuzz(15))









```
